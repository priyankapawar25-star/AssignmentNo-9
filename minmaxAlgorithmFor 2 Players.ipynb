```python
import math
#Board cells: empty = ' ',player 'x', opponent 'o'
board = [' ' for _ in range(9)]

def print_board():
  for i in range(3):
    print(board[3*i], '|', board[3*i+1],'|', board[3*i+2])
    if i < 2:
      print('---------')

def is_winner(b,player):
  #Winnng combinations in Tic-Tac-Toe
  wins = [
      (0,1,2),(3,4,5),(6,7,8), #rows
      (0,3,6),(1,4,7),(2,5,8), #columns
      (0,4,8),(2,4,6),         #digonals
  ]
  for a,b_,c in wins:
    if b[a] == b[b_] == b[c] == player: # Fixed win condition to check all three positions
      return True
  return False

def is_board_full(b):
  return ' ' not in b

def minimax(b, depth, is_maximizing):
  """
  Minimax recursive function:
  - is_maximizing:True if the current turn is maximizer (x), else minimizer(o).
  - Returns the best score from the perspective of the maximizer.
  """
  if is_winner(b,'x'):
    return 10 - depth # maximizer wins
  if is_winner(b,'o'):
    return depth - 10 # minimizer wins
  if is_board_full(b):
    return 0 # Draw

  if is_maximizing:
    best_score = -math.inf
    for i in range (9):
      if b[i] == ' ':
        b[i] = 'x' # Try this move
        score = minimax(b, depth + 1, False)
        b[i] = ' ' # Undo the move
        best_score = max(score, best_score)
    return best_score
  else:
    best_score = math.inf
    for i in range(9):
      if b[i] == ' ':
        b[i] = 'o' # Try this move
        score = minimax(b, depth + 1, True)
        b[i] = ' ' # Undo the move
        best_score = min(score, best_score)
    return best_score

def best_move():
  """
  Find the best for the maximizer (x).
  """
  best_score = -math.inf
  move = -1
  for i in range(9):
    if board[i] == ' ':
      board[i] = 'x'
      score = minimax(board,0, False)
      board[i] = ' '
      if score > best_score: # Added colon and updated best_score
        best_score = score
        move = i
  return move # Corrected indentation

def play_game():
  """
  Simple game loop: Human (o) vs AI (X using minimax).
  """
  print("Board positions are numbered 0 to 8:")
  print("0|1|2")
  print("-----")
  print("3|4|5")
  print("-----")
  print("6|7|8")
  print()

  while True:
    # AI move
    move = best_move()
    if move == -1: # Fixed '--' to '=='
      print("Game Over! It is a tie.")
      break
    board[move] = 'x'
    print("AI (x) plays move:", move)
    print_board()
    if is_winner(board, 'x'):
      print("AI (X) wins!")
      break
    if is_board_full(board): # Moved outside the break and corrected indentation
      print("Game Over!It's a tie.")
      break

    #Human move
    while True:
      try:
        human_move = int(input("Enter your move(0-8):"))
        if 0 <= human_move <= 8 and board[human_move] == ' ': # Fixed `human move` typo
          board[human_move] = 'o'
          break
        else:
          print("Invalid move. Try again.")
      except ValueError:
        print("Please enter a valid number.") # Corrected indentation and typo "enetr"

    print_board()
    if is_winner(board, 'o'):
      print("You win!")
      break
    if is_board_full(board):
      print("Game Over! It's a tie.")
      break

if __name__ == "__main__": # Fixed typo 'if__name__'
  play_game()
```
